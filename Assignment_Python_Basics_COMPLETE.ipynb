````python
# Convert the whole string to lower case.
print(sentence.lower())
````

````python
# Extract a substring from the 10th character from left to the 10th character from right.
print(sentence[9:-9])
````

````python
# Split the sentence into a list of substrings by space ' '.
print(sentence.split())
````

````python
# Given the sentence, output a list of words in lowercase (without symbols such as '.' and '-').
print(sentence.lower().replace('. ', ' ').replace(' -', ''))
````

````python
# Count the total number of vowels in the sentence.
vowels = ['a','e','i','o','u']
#vowels = 'aeiou'
total = 0
for v in vowels:
  total += sentence.lower().count(v)
print(total)
````

````python
# Find the positions (indexes) of all vowels in the sentence.
positions = []
for i in range(len(sentence)):
    if sentence[i].lower() in vowels:
      positions.append(i)
print(positions)
````
