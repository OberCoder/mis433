## Part 2: Data Summarization

```python
# Total "Gross" of all top 1000 movies
imdb['Gross'].sum()
```

```python
# Average "No_of_Votes" of all movies
imdb['No_of_Votes'].mean()
```

```python
# Count movies in each decade (e.g., ..., 1980, 1990, 2000, 2010, 2020)
# Sort decades by the number of movies in descending order
imdb['Decade'].value_counts()
```

```python
# Count movies by different directors.
# Show the top 10 directors with the most movies in this list
imdb['Director'].value_counts().head(10)
```

```python
# For each director, calculate the number of movies, average IMDB_Rating, and total gross.
# Rank them in descending order of total gross
# Show the top 10 directors with the highest total gross
imdb.groupby('Director').agg(
    Num_Movies=('Series_Title', 'count'),
    Avg_Rating=('IMDB_Rating', 'mean'),
    Total_Gross=('Gross', 'sum')
).sort_values('Total_Gross', ascending=False).head(10)
```

```python
# Group movies by decade and director
# In each group (i.e., for each decade and each director),
# calculate the number of movies and average IMDB rating.
# Sort in descending order of movie count
imdb.groupby(['Decade', 'Director']).agg(
    Num_Movies=('Series_Title', 'count'),
    Avg_Rating=('IMDB_Rating', 'mean')
).sort_values('Num_Movies', ascending=False)
```

```python
# Find the top 3 highest rated movies of each year since 2010.
# Show their released year, ranking, title, and IMDB_Rating.
# Sort them in descending order of year and ascending order of ranking
recent = imdb[imdb['Released_Year'] >= 2010].copy()
recent['Ranking'] = recent.groupby('Released_Year')['IMDB_Rating'].rank(method='first', ascending=False).astype(int)
top3 = recent[recent['Ranking'] <= 3]
top3[['Released_Year', 'Ranking', 'Series_Title', 'IMDB_Rating']].sort_values(['Released_Year', 'Ranking'], ascending=[False, True])
```

```python
# Find all directors whose movies appeared in at least five different decades.
# Your result should include: director, decade, and the number of movies in the decade.
dd = imdb.groupby(['Director', 'Decade']).size().reset_index(name='Num_Movies')
dd['Num_Decades'] = dd.groupby('Director')['Decade'].transform('count')
dd[dd['Num_Decades'] >= 5][['Director', 'Decade', 'Num_Movies']]
```

```python
# Ask two questions about this dataset and write your code to answer them.
# Question 1: Which certificate ratings have the highest average gross, and how many movies does each include?
imdb.groupby('Certificate').agg(
    Num_Movies=('Series_Title', 'count'),
    Avg_Gross=('Gross', 'mean')
).sort_values('Avg_Gross', ascending=False)
```

```python
# Question 2: # Question 2: Does a longer runtime go with a higher IMDB rating?
imdb['Runtime_Group'] = pd.cut(imdb['Runtime_min'], bins=[0, 90, 120, 150, 500], labels=['Under 90', '90-120', '120-150', 'Over 150'])
imdb.groupby('Runtime_Group').agg(
    Num_Movies=('Series_Title', 'count'),
    Avg_Rating=('IMDB_Rating', 'mean')
)
```
